```python

uk_orders = order_df[
    order_df['Country'] == 'United Kingdom'
]['TotalSales']

other_orders = order_df[
    order_df['Country'] != 'United Kingdom'
]['TotalSales']

print("UK Orders:", len(uk_orders))
print("Non-UK Orders:", len(other_orders))

print("UK Average Order Value:", round(uk_orders.mean(), 2))
print("Non-UK Average Order Value:", round(other_orders.mean(), 2))
```

```python

from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    uk_orders,
    other_orders,
    equal_var=False
)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a statistically significant difference.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is insufficient evidence of a difference.")
```

```python

from scipy.stats import t

import numpy as np

mean_difference = uk_orders.mean() - other_orders.mean()

se = np.sqrt(
    uk_orders.var(ddof=1) / len(uk_orders) +
    other_orders.var(ddof=1) / len(other_orders)
)

df_welch = (
    (uk_orders.var(ddof=1)/len(uk_orders) +
     other_orders.var(ddof=1)/len(other_orders))**2
    /
    (
        (uk_orders.var(ddof=1)/len(uk_orders))**2 /
        (len(uk_orders)-1)
        +
        (other_orders.var(ddof=1)/len(other_orders))**2 /
        (len(other_orders)-1)
    )
)

critical_value = t.ppf(0.975, df_welch)

lower = mean_difference - critical_value * se
upper = mean_difference + critical_value * se

print("Mean Difference:", round(mean_difference, 2))
print("95% Confidence Interval:")
print(round(lower, 2), "to", round(upper, 2))
```

```python
import pandas as pd
import numpy as np
import requests
import os

# Download dataset if not available
file_path = "online_retail_II.xlsx"

if not os.path.exists(file_path):
    url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00502/online_retail_II.xlsx"
    response = requests.get(url)
    response.raise_for_status()

    with open(file_path, "wb") as f:
        f.write(response.content)

# Load dataset
df = pd.read_excel(file_path)

# Clean dataset
clean_df = df.copy()
clean_df.columns = clean_df.columns.str.strip()

clean_df.rename(columns={
    'Customer ID': 'CustomerID',
    'Invoice': 'InvoiceNo',
    'Price': 'UnitPrice'
}, inplace=True)

clean_df['Description'] = clean_df['Description'].fillna('Unknown Product')
clean_df = clean_df.drop_duplicates().copy()

# Create model dataset
model_df = clean_df.copy()

model_df['InvoiceDate'] = pd.to_datetime(
    model_df['InvoiceDate'], errors='coerce'
)

model_df['UnitPrice'] = pd.to_numeric(
    model_df['UnitPrice'], errors='coerce'
)

model_df['Quantity'] = pd.to_numeric(
    model_df['Quantity'], errors='coerce'
)

model_df = model_df.dropna(
    subset=['InvoiceNo', 'InvoiceDate', 'UnitPrice',
            'Quantity', 'Country']
).copy()

model_df['InvoiceNo'] = model_df['InvoiceNo'].astype(str)

model_df['Cancellation'] = (
    model_df['InvoiceNo'].str.startswith('C')
).astype(int)

model_df['QuantityAbs'] = model_df['Quantity'].abs()
model_df['Month'] = model_df['InvoiceDate'].dt.month
model_df['Weekday'] = model_df['InvoiceDate'].dt.dayofweek
model_df['Hour'] = model_df['InvoiceDate'].dt.hour

print("Dataset loaded and model dataset created!")
print("Original dataset:", df.shape)
print("Clean dataset:", clean_df.shape)
print("Model dataset:", model_df.shape)
print("\nCancellation counts:")
print(model_df['Cancellation'].value_counts())
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Select input features
X = model_df[
    ['QuantityAbs', 'UnitPrice', 'Month',
     'Weekday', 'Hour', 'Country']
].copy()

# Select target
y = model_df['Cancellation']

# Group by invoice to avoid splitting the same invoice
groups = model_df['InvoiceNo'].astype(str)

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

numeric_features = [
    'QuantityAbs', 'UnitPrice', 'Month', 'Weekday', 'Hour'
]

categorical_features = ['Country']

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

print("Preprocessor created successfully!")
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        max_iter=1000,
        class_weight='balanced'
    ))
])

logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
from sklearn.ensemble import RandomForestClassifier

random_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=30,
        max_depth=12,
        min_samples_leaf=5,
        random_state=42,
        class_weight='balanced',
        n_jobs=-1
    ))
])

random_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
sample_size = min(100000, len(X_train))

X_sample = X_train.sample(
    n=sample_size,
    random_state=42
)

y_sample = y_train.loc[X_sample.index]

random_model.fit(X_sample, y_sample)

print("Random Forest trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Evaluation libraries imported!")
```

```python
logistic_pred = logistic_model.predict(X_test)

random_pred = random_model.predict(X_test)

print("Predictions completed!")
```

```python
results = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest'],
    'Accuracy': [
        accuracy_score(y_test, logistic_pred),
        accuracy_score(y_test, random_pred)
    ],
    'Precision': [
        precision_score(y_test, logistic_pred, zero_division=0),
        precision_score(y_test, random_pred, zero_division=0)
    ],
    'Recall': [
        recall_score(y_test, logistic_pred, zero_division=0),
        recall_score(y_test, random_pred, zero_division=0)
    ],
    'F1 Score': [
        f1_score(y_test, logistic_pred, zero_division=0),
        f1_score(y_test, random_pred, zero_division=0)
    ]
})

display(results.round(3))
```

```python
sales_df = clean_df.copy()

sales_df['InvoiceDate'] = pd.to_datetime(
    sales_df['InvoiceDate'], errors='coerce'
)

sales_df['Quantity'] = pd.to_numeric(
    sales_df['Quantity'], errors='coerce'
)

sales_df['UnitPrice'] = pd.to_numeric(
    sales_df['UnitPrice'], errors='coerce'
)

sales_df = sales_df.dropna(
    subset=['CustomerID', 'InvoiceDate', 'Quantity', 'UnitPrice']
)

sales_df = sales_df[
    (sales_df['Quantity'] > 0) &
    (sales_df['UnitPrice'] > 0)
].copy()

sales_df['TotalSales'] = (
    sales_df['Quantity'] * sales_df['UnitPrice']
)

print("Customer sales data prepared!")
print(sales_df.shape)
```

```python
snapshot_date = sales_df['InvoiceDate'].max() + pd.Timedelta(days=1)

rfm = sales_df.groupby('CustomerID').agg(
    Recency=('InvoiceDate', lambda x: (snapshot_date - x.max()).days),
    Frequency=('InvoiceNo', 'nunique'),
    Monetary=('TotalSales', 'sum')
).reset_index()

display(rfm.head())

print("Total customers:", len(rfm))
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm['RFM_Score'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)

display(rfm.head(10))
```

```python
def customer_segment(row):
    if row['R_Score'] >= 3 and row['F_Score'] >= 3:
        return 'Loyal Customers'
    elif row['R_Score'] == 4:
        return 'Recent Customers'
    elif row['R_Score'] == 1:
        return 'At Risk Customers'
    else:
        return 'Regular Customers'

rfm['Segment'] = rfm.apply(customer_segment, axis=1)

print(rfm['Segment'].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.countplot(
    data=rfm,
    x='Segment',
    order=rfm['Segment'].value_counts().index
)

plt.title('Customer Segmentation')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=20)
plt.show()
```

```python
segment_revenue = rfm.groupby('Segment')['Monetary'].sum().sort_values(
    ascending=False
)

plt.figure(figsize=(9, 5))

segment_revenue.plot(kind='bar')

plt.title('Revenue by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Revenue')
plt.xticks(rotation=20)
plt.show()
```

```python
segment_summary = rfm.groupby('Segment').agg(
    Customer_Count=('CustomerID', 'count'),
    Average_Recency=('Recency', 'mean'),
    Average_Frequency=('Frequency', 'mean'),
    Average_Spending=('Monetary', 'mean'),
    Total_Revenue=('Monetary', 'sum')
).round(2)

display(segment_summary)
```

```python
daily_sales = sales_df.copy()

daily_sales['Date'] = daily_sales['InvoiceDate'].dt.floor('D')

daily_sales = daily_sales.groupby('Date')['TotalSales'].sum()

daily_sales = daily_sales.asfreq('D', fill_value=0)

daily_sales = daily_sales.reset_index()

daily_sales.rename(columns={'TotalSales': 'Sales'}, inplace=True)

print("Daily sales data prepared!")
display(daily_sales.head())
```

```python
plt.figure(figsize=(14, 5))

plt.plot(daily_sales['Date'], daily_sales['Sales'])

plt.title('Daily Retail Sales Trend')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
daily_sales['Day'] = daily_sales['Date'].dt.day
daily_sales['Month'] = daily_sales['Date'].dt.month
daily_sales['Weekday'] = daily_sales['Date'].dt.dayofweek
daily_sales['Year'] = daily_sales['Date'].dt.year

daily_sales['Lag_1'] = daily_sales['Sales'].shift(1)
daily_sales['Lag_7'] = daily_sales['Sales'].shift(7)

daily_sales['Rolling_7'] = (
    daily_sales['Sales'].shift(1).rolling(7).mean()
)

daily_sales = daily_sales.dropna().reset_index(drop=True)

print("Time-based features created!")
display(daily_sales.head())
```

```python
split_index = int(len(daily_sales) * 0.8)

train_data = daily_sales.iloc[:split_index]
test_data = daily_sales.iloc[split_index:]

features = [
    'Day', 'Month', 'Weekday', 'Year',
    'Lag_1', 'Lag_7', 'Rolling_7'
]

X_train_ts = train_data[features]
y_train_ts = train_data['Sales']

X_test_ts = test_data[features]
y_test_ts = test_data['Sales']

print("Training rows:", len(X_train_ts))
print("Testing rows:", len(X_test_ts))
```

```python
from sklearn.ensemble import RandomForestRegressor

forecast_model = RandomForestRegressor(
    n_estimators=50,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

forecast_model.fit(X_train_ts, y_train_ts)

print("Sales forecasting model trained!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

forecast_predictions = forecast_model.predict(X_test_ts)

mae = mean_absolute_error(y_test_ts, forecast_predictions)

rmse = np.sqrt(
    mean_squared_error(y_test_ts, forecast_predictions)
)

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    test_data['Date'],
    y_test_ts,
    label='Actual Sales'
)

plt.plot(
    test_data['Date'],
    forecast_predictions,
    label='Predicted Sales'
)

plt.title('Actual vs Predicted Daily Sales')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```
